## Creating the function

```python
# RUSHING
def get_carries(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['carries'].values[0]

def get_receiving_yards(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['receiving_yards'].values[0]

def get_rushing_tds(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['rushing_tds'].values[0]
```

```python
# PASSING
def get_completions(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['completions'].values[0]

def get_passing_yards(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['passing_yards'].values[0]

def get_passing_tds(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['passing_tds'].values[0]
```

```python
# RECEIVING
def get_receptions(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['receptions'].values[0]

def get_rushing_yards(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['rushing_yards'].values[0]

def get_receiving_tds(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['receiving_tds'].values[0]
```

```python
# TURNOVERS
def get_fumbles(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        total_fumbles = (player_data['sack_fumbles_lost'].values[0] 
                        + player_data['rushing_fumbles_lost'].values[0] 
                        + player_data['receiving_fumbles_lost'].values[0])
        return total_fumbles

def get_interceptions(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['interceptions'].values[0]
    
def get_turnovers(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        total_fumbles = (player_data['sack_fumbles_lost'].values[0] 
                        + player_data['rushing_fumbles_lost'].values[0] 
                        + player_data['receiving_fumbles_lost'].values[0])
        return total_fumbles + player_data['interceptions'].values[0]

def games_played(data, player_id, season):
    player_data = data[(data['player_id'] == player_id) & (data['season'] == season)]
    if player_data.empty:
        #print(f"\tERROR: No data found for player_id {player_id} in season {season}.")
        return "--"
    else:
        return player_data['week'].values[0]
```

## Applying it to adp files

```python
years = list(range(2014, 2025))
weekly = nfl.import_weekly_data(years)
weekly = weekly[weekly['week']>=1]
weekly = weekly[((weekly['season']<=2020) & (weekly['week']<=17)) |
                ((weekly['season']>2020) & (weekly['week']<=18))]
agrupado = weekly.groupby(['player_id','player_display_name','season']).agg({
    'week': 'count',
    'completions': 'sum',
    'passing_yards':'sum',
    'passing_tds':'sum',
    'sack_fumbles_lost':'sum',
    'interceptions':'sum',
    'carries': 'sum',
    'rushing_yards': 'sum',
    'rushing_tds': 'sum',
    'rushing_fumbles_lost':'sum',
    'receptions': 'sum',
    'receiving_yards': 'sum',
    'receiving_tds': 'sum',
    'receiving_fumbles_lost': 'sum'
})
agrupado = agrupado.reset_index()
```

```python
years = list(range(2015, 2026))
#years = [2025]
for year in years:
    filepath = f"adp_full_gsispos_{year}.csv"
    adp_data = pd.read_csv(filepath)
    adp_data['prev_year_games'] = adp_data.apply(
        lambda row: games_played(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    #Passing
    adp_data['prev_year_completions'] = adp_data.apply(
        lambda row: get_completions(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    adp_data['prev_year_passing_yards'] = adp_data.apply(
        lambda row: get_passing_yards(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    adp_data['prev_year_passing_tds'] = adp_data.apply(
        lambda row: get_passing_tds(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    #Rushing
    adp_data['prev_year_carries'] = adp_data.apply(
        lambda row: get_carries(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    adp_data['prev_year_rushing_yards'] = adp_data.apply(
        lambda row: get_rushing_yards(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    adp_data['prev_year_rushing_tds'] = adp_data.apply(
        lambda row: get_rushing_tds(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    #Receiving
    adp_data['prev_year_receptions'] = adp_data.apply(
        lambda row: get_receptions(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    adp_data['prev_year_receiving_yards'] = adp_data.apply(
        lambda row: get_receiving_yards(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    adp_data['prev_year_receiving_tds'] = adp_data.apply(
        lambda row: get_receiving_tds(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    #Turnovers
    adp_data['prev_year_turnovers'] = adp_data.apply(
        lambda row: get_turnovers(agrupado, row['gsis_id'], year-1),
        axis=1
    )
    if year < 2025:
        #Passing
        adp_data['obj_completions'] = adp_data.apply(
            lambda row: get_completions(agrupado, row['gsis_id'], year),
            axis=1
        )
        adp_data['obj_passing_yards'] = adp_data.apply(
            lambda row: get_passing_yards(agrupado, row['gsis_id'], year),
            axis=1
        )
        adp_data['obj_passing_tds'] = adp_data.apply(
            lambda row: get_passing_tds(agrupado, row['gsis_id'], year),
            axis=1
        )
        #Rushing
        adp_data['obj_carries'] = adp_data.apply(
            lambda row: get_carries(agrupado, row['gsis_id'], year),
            axis=1
        )
        adp_data['obj_rushing_yards'] = adp_data.apply(
            lambda row: get_rushing_yards(agrupado, row['gsis_id'], year),
            axis=1
        )
        adp_data['obj_rushing_tds'] = adp_data.apply(
            lambda row: get_rushing_tds(agrupado, row['gsis_id'], year),
            axis=1
        )
        #Receiving
        adp_data['obj_receptions'] = adp_data.apply(
            lambda row: get_receptions(agrupado, row['gsis_id'], year),
            axis=1
        )
        adp_data['obj_receiving_yards'] = adp_data.apply(
            lambda row: get_receiving_yards(agrupado, row['gsis_id'], year),
            axis=1
        )
        adp_data['obj_receiving_tds'] = adp_data.apply(
            lambda row: get_receiving_tds(agrupado, row['gsis_id'], year),
            axis=1
        )
        #Turnovers
        adp_data['obj_turnovers'] = adp_data.apply(
            lambda row: get_turnovers(agrupado, row['gsis_id'], year),
            axis=1
        )
    
    adp_data.to_csv(f"adp_full_{year}_with_prev_obj.csv", index=False)
    print(f"Finished {year}.")

print(adp_data.columns)
```
