```python
import pandas as pd
```

```python
player_stats = pd.read_csv("player_weekly_data_2014_24.csv")
offense_stats = pd.read_csv("team_offense_2014_24.csv")

lj = pd.merge(player_stats, offense_stats, how='left', on = ['game_id', 'ball_team'])
print(lj.columns)
metrics = ['completions', 'attempts', 'passing_yards', 'passing_tds', 
           'carries', 'rushing_yards', 'rushing_tds',
           'receptions', 'targets', 'receiving_yards', 'receiving_tds']
for met in metrics:
    lj[f"{met}_share"] = lj[f"{met}_x"]/lj[f"{met}_y"]
share_fil = lj[['season', 'week', 'game_id', 'ball_team','opponent_team',
                'player_id', 'player_display_name', 'position', 
                'completions_share', 'attempts_share', 'passing_yards_share', 'passing_tds_share', 
                'carries_share', 'rushing_yards_share', 'rushing_tds_share',
                'receptions_share', 'targets_share', 'receiving_yards_share', 'receiving_tds_share']]
share_fil = share_fil.fillna(0)
tracy = share_fil[share_fil['player_id'] == '00-0039910'] #'00-0039384']
tracy.head(10)
```
