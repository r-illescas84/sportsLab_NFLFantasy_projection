```python
import pandas as pd
import numpy as np
```

```python
weekly_fil = pd.read_csv('player_weekly_data_2014_24.csv')
weekly_fil = weekly_fil.sort_values(["player_id", "season", "week"])

cols = ["completions", "attempts", "passing_yards", "passing_tds", "interceptions",'carries', 'rushing_yards',
       'rushing_tds', 'receptions', 'targets', 'receiving_yards',
       'receiving_tds', 'fumbles_lost']
g = weekly_fil.groupby(["player_id", "season"], sort=False)
prev_sum = g[cols].cumsum().shift(1)
prev_cnt = g.cumcount()
season_avg = prev_sum.div(prev_cnt.replace(0, np.nan), axis=0)
season_avg.columns = [f"{c}_season_avg" for c in cols]
weekly_fil = pd.concat([weekly_fil, season_avg], axis=1)
print(weekly_fil.columns)
tt = weekly_fil[(weekly_fil['player_id'] == '00-0026498') & (weekly_fil['season']==2019)] #'00-0039384']
tt[['season','week', 'player_display_name', 'passing_yards', 'passing_yards_season_avg', 'rn']].head(10)
weekly_fil = weekly_fil[['player_id', 'recent_team', 'player_display_name', 'position', 'season',
       'week', 'opponent_team', 'game_id', 'ball_team', 'rn',
       'completions_season_avg', 'attempts_season_avg', 'passing_yards_season_avg',
       'passing_tds_season_avg', 'interceptions_season_avg', 'carries_season_avg',
       'rushing_yards_season_avg', 'rushing_tds_season_avg', 'receptions_season_avg',
       'targets_season_avg', 'receiving_yards_season_avg', 'receiving_tds_season_avg',
       'fumbles_lost_season_avg']]
weekly_fil.to_csv('player_season_avg_2014-24.csv', index=False)
```

```python
def_stats = pd.read_csv("team_defense_2014_24.csv")
def_stats = def_stats.sort_values(["opponent_team", "rn"])
def_stats['season'] = def_stats['game_id'].str[:4].astype(int)

cols = ["passing_yards", "passing_tds", "interceptions", 'rushing_yards',
       'rushing_tds', 'receiving_yards',
       'receiving_tds', 'fumbles_lost']

g = def_stats.groupby(["opponent_team", "season"], sort=False)
prev_sum = g[cols].cumsum().shift(1)
prev_cnt = g.cumcount()
season_avg = prev_sum.div(prev_cnt.replace(0, np.nan), axis=0)
season_avg.columns = [f"{c}_season_avg" for c in cols]
def_stats = pd.concat([def_stats, season_avg], axis=1)
print(def_stats.columns)
tt = def_stats[(def_stats['opponent_team'] == 'LA') & (def_stats['season']==2019)] #'00-0039384']
tt[['season', 'passing_yards', 'passing_yards_season_avg', 'rn']].head(10)
def_stats = def_stats[['opponent_team', 'season',
       'game_id', 'rn',
       'passing_yards_season_avg',
       'passing_tds_season_avg', 'interceptions_season_avg', 
       'rushing_yards_season_avg', 'rushing_tds_season_avg', 
       'receiving_yards_season_avg', 'receiving_tds_season_avg',
       'fumbles_lost_season_avg']]
def_stats.to_csv('team_defense_season_avg_2014-24.csv', index=False)
```

```python
off_stats = pd.read_csv("team_offense_2014_24.csv")
off_stats = off_stats.sort_values(["ball_team", "rn"])
off_stats['season'] = off_stats['game_id'].str[:4].astype(int)

cols = ["passing_yards", "passing_tds", 'rushing_yards', 
       'carries', 'attempts', 'targets', 'receptions', 'completions',
       'rushing_tds', 'receiving_yards',
       'receiving_tds', 'carries', 'completions', 'attempts', 'receptions']
g = off_stats.groupby(["ball_team", "season"], sort=False)
prev_sum = g[cols].cumsum().shift(1)
prev_cnt = g.cumcount()
season_avg = prev_sum.div(prev_cnt.replace(0, np.nan), axis=0)
season_avg.columns = [f"{c}_season_avg" for c in cols]
off_stats = pd.concat([off_stats, season_avg], axis=1)
print(off_stats.columns)
tt = off_stats[(off_stats['ball_team'] == 'LA') & (off_stats['season']==2019)] #'00-0039384']
tt[['season', 'ball_team', 'passing_yards', 'passing_yards_season_avg', 'rn']].head(10)
off_stats = off_stats[['ball_team', 'season',
       'game_id', 'rn',
       'completions_season_avg', 'attempts_season_avg', 'passing_yards_season_avg',
       'passing_tds_season_avg', 'carries_season_avg',
       'rushing_yards_season_avg', 'rushing_tds_season_avg', 'receptions_season_avg',
       'targets_season_avg', 'receiving_yards_season_avg', 'receiving_tds_season_avg']]
off_stats.to_csv('team_offense_season_avg_2014-24.csv', index=False)
```
