```python
import pandas as pd
```

```python
weekly_fil = pd.read_csv('../data/player_weekly_data_2014_24.csv')
weekly_fil = weekly_fil.sort_values(["player_id", "rn"])

cols = ["completions", "attempts", "passing_yards", "passing_tds", "interceptions",'carries', 'rushing_yards',
       'rushing_tds', 'receptions', 'targets', 'receiving_yards',
       'receiving_tds', 'fumbles_lost']
weekly_fil[[f"{c}_last5" for c in cols]] = (
    weekly_fil
    .groupby("player_id")[cols]
    .apply(lambda g: g.shift().rolling(5, min_periods=1).mean())
    .reset_index(drop=True)
)
print(weekly_fil.columns)
tt = weekly_fil[weekly_fil['player_id'] == '00-0039910'] #'00-0039384']
tt[['season','week', 'player_display_name', 'passing_yards', 'passing_yards_last5', 'rn']].head(10)
weekly_fil = weekly_fil[['player_id', 'recent_team', 'player_display_name', 'position', 'season',
       'week', 'opponent_team', 'game_id', 'ball_team', 'rn',
       'completions_last5', 'attempts_last5', 'passing_yards_last5',
       'passing_tds_last5', 'interceptions_last5', 'carries_last5',
       'rushing_yards_last5', 'rushing_tds_last5', 'receptions_last5',
       'targets_last5', 'receiving_yards_last5', 'receiving_tds_last5',
       'fumbles_lost_last5']]
weekly_fil.to_csv('../data/player_last5_avg_2014-24.csv', index=False)
```

```python
def_stats = pd.read_csv("../data/team_defense_2014_24.csv")
def_stats = def_stats.sort_values(["opponent_team", "rn"])

cols = ["passing_yards", "passing_tds", "interceptions", 'rushing_yards',
       'rushing_tds', 'receiving_yards',
       'receiving_tds', 'fumbles_lost']
def_stats[[f"{c}_last5" for c in cols]] = (
    def_stats
    .groupby("opponent_team")[cols]
    .apply(lambda g: g.shift().rolling(5, min_periods=1).mean())
)
print(def_stats.columns)
tt = def_stats[def_stats['opponent_team'] == 'LA']
tt[['game_id','opponent_team', 'passing_yards', 'passing_yards_last5', 'rn']].head(10)
def_stats.to_csv('../data/team_defense_last5_avg_2014_24.csv', index=False)
```

```python
off_stats = pd.read_csv("../data/team_offense_2014_24.csv")
off_stats = off_stats.sort_values(["ball_team", "rn"])

cols = ["passing_yards", "passing_tds", 'rushing_yards',
       'rushing_tds', 'receiving_yards',
       'receiving_tds', 'carries', 'completions', 'attempts', 'receptions']
off_stats[[f"{c}_last5" for c in cols]] = (
    off_stats
    .groupby("ball_team")[cols]
    .apply(lambda g: g.shift().rolling(5, min_periods=1).mean())
)
print(off_stats.columns)
tt = off_stats[off_stats['ball_team'] == 'LA']
tt.head(10)
off_stats.to_csv('../data/team_offense_last5_avg_2014_24.csv', index=False)
```
