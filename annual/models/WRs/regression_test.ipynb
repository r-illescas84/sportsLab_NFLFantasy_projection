```python
import pandas as pd
```

```python
first = 2016
last = 2024
for year in list(range(first, last+1)):
    print("---Ricky---")
    ricky = pd.read_csv(f"../../WR/added_ricky_metrics_{year}.csv") 
    print(len(ricky))
    ricky = ricky[ricky['season']+1 != ricky['rookie_season']]
    print(len(ricky))
    cols_to_drop = ['Bye', 'POS', 'Yahoo', 'Sleeper', 'RTSports', 'ESPN', 'NFL','FFC','Fantrax','rookie_season']
    print(ricky.columns)
    ricky = ricky.drop(columns=[col for col in cols_to_drop if col in ricky.columns])

    print("---Emi---")
    emi = pd.read_csv(f"../../WR/efficiencyWR_{year-1}.csv")
    print(len(emi))
    emi = emi[['player_id', 'games_played', 'total_player_yards', 'avg_player_yards', 'avg_def_difficulty', 'avg_exploitation']]
    rickyEmi = pd.merge(left=ricky, right=emi, how='inner', left_on='gsis_id', right_on='player_id')
    print(len(rickyEmi))

    print("---Emi2---")
    emi2 = pd.read_csv(f"../../WR/redzone_statsWR_{year-1}.csv")
    print(len(emi2))
    emi2 = emi2[['player_id','redzone_touches','redzone_yards','redzone_tds']]
    rickyEmi = pd.merge(left=rickyEmi, right=emi2, how='inner', left_on='gsis_id', right_on='player_id')
    print(len(rickyEmi))

    print("---Zama---")
    if year==2025:
        zama = pd.read_csv(f"../../WR/wr_zama{year-2000}.csv", sep=';')
    else:
        zama = pd.read_csv(f"../../WR/wr_completo{year-2000}.csv")
    zama = zama[['gsis_id','targets','recepciones', 'rec_yards', 'rec_ypg', 'rec_20plus',
        'rec_td', 'attempts', 'rush_yards', 'rush_ypg', 'rush_td',
        'fumbles_perdidos', 'total_touchdowns', 'partidos']]
    print(len(zama))
    rickyEmiZama = pd.merge(left=rickyEmi, right=zama, how='inner', on='gsis_id')
    print(len(rickyEmiZama))

    print("---Alex---")
    alex = pd.read_csv(f"../../WR/wr_target_share_{year-1}.csv")
    print(len(alex))
    alex = alex[['gsis_id','targets_totales','targets_por_partido','targets_por_partido_equipo', 'posicion', 'equipo', 'target_share_pct']]
    rickyEmiZamaAlex = pd.merge(left=rickyEmiZama, right=alex, how='inner', left_on='gsis_id', right_on='gsis_id')
    print(len(rickyEmiZamaAlex))

    if year==2016:
        data = rickyEmiZamaAlex
    else:
        data = pd.concat([data, rickyEmiZamaAlex], ignore_index=True)

data.to_csv(f"complete_{first}_{last}.csv", index=False)
```
